```python
#palindrome string
s="racecary"
if s== s[::-1]:
    print("Palindrome using slicing")
else:
    print("Not a palindrome")

#palindrome using loop
flag=True

for i in range(len(s)//2):
    if s[i]!= s[len(s)-1-i]:
        flag=False
        break;
if flag:
    print("Palindrome")
else:
    print("Not a palindrome")
    
```
