```python
#part c axis aggregations np.where,np.select
spend = np.array([
    [20, 35,   0, 15, 30],   # customer 0: regular
    [ 0,  0, 500,  0,  0],   # customer 1: one huge spend, nothing else
    [12, 10,  14, 11, 13],   # customer 2: small and steady
    [60, 45,  80, 55, 70],   # customer 3: big and steady
], dtype=float)
print(spend.shape, spend.ndim, spend.size)   # (4, 5) 2 20

spend.sum(), spend.mean(), spend.min(), spend.max()   # 970.0  48.5  0.0  500.0
np.median(spend)                                      # 14.5
np.percentile(spend, [25, 75])                        # [ 7.5 47.5]   Q1 and Q3
```

```python
''' 
axis=0 collapses the rows. You move down each column, so you get one value per column (per day).
axis=1 collapses the columns. You move across each row, so you get one value per row (per customer).
'''
print(spend.sum(axis=0))     # [ 92.  90. 594.  81. 113.]   shape (5,)
print(spend.mean(axis=0))    # [ 23.  22.5 148.5 20.25 28.25]


spend.sum(axis=1)                  # [100. 500.  60. 310.]   shape (4,)
spend.mean(axis=1)                 # [ 20. 100.  12.  62.]
spend.std(axis=1, ddof=1)          # [ 13.69 223.61  1.58  13.51]
np.median(spend, axis=1)           # [20.  0. 12. 60.]
np.percentile(spend, 50, axis=1)   # [20.  0. 12. 60.]   same as median


#argmax:argmin : position of max and min
print(spend.max(axis=1))      # [ 35. 500.  14.  80.]
print(spend.argmax(axis=1))   # [1 2 2 2]   which day each customer spent the most
```

```python
#keepdims=True keeps the collapsed dimension as size 1, so the result lines up:
tot_k = spend.sum(axis=1, keepdims=True)   # shape (4, 1)
share = spend / tot_k
# [[0.2  0.35 0.   0.15 0.3 ]
#  [0.   0.   1.   0.   0.  ]
#  [0.2  0.17 0.23 0.18 0.22]
#  [0.19 0.15 0.26 0.18 0.23]]
share.sum(axis=1)                          # [1. 1. 1. 1.]  ✔
```

```python
#safe division with np.where
''' np.errstate(...)	with block that temporarily silences numpy warnings (here, divide-by-zero)
np.zeros_like(arr)	New array of zeros with the same shape and dtype as arr (also np.ones_like, np.full_like)
np.divide(a, b, out=, where=)	Division that only computes where where is True and leaves out untouched elsewhere'''
num = np.array([10., 0., 5.]); den = np.array([2., 0., 0.])

with np.errstate(divide="ignore", invalid="ignore"):
    ratio = np.where(den > 0, num / den, 0.0)        # [5. 0. 0.]

np.divide(num, den, out=np.zeros_like(num), where=den > 0)   # [5. 0. 0.]
```

```python
#4d: Outlier detection in pure numpy

#method 1 zscore
def zscore_outliers(x, threshold=3.0):
    x = np.asarray(x, dtype=float)
    mu = np.nanmean(x)
    sd = np.nanstd(x, ddof=1)
    if sd == 0 or np.isnan(sd):
        return np.zeros(x.shape, dtype=bool)
    z = (x - mu) / sd
    return np.abs(z) > threshold

#IQR methd
def iqr_outliers(x, k=1.5):
    x = np.asarray(x, dtype=float)
    q1, q3 = np.nanpercentile(x, [25, 75])
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    return (x < lower) | (x > upper)
```
