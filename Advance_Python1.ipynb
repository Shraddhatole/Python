```python
#generator comprehension
import sys
sqr=(x**2 for x in range(100))
sqr_lst=[x**2 for x in range(100)]
print(sqr) #op:<generator object <genexpr> at 0x11168f780>
print(sys.getsizeof(sqr))  # 208
print(sys.getsizeof(sqr_lst)) #920
next(sqr)
next(sqr)
```

```python
#Closure : when inner function still remebers the variable from outer func even after execution of outer is completed

def outer():
    message = "Hello"

    def inner():
        print(message)

    return inner


my_function = outer()
my_function()
```

```python
#decorator:A decorator is a function that adds extra functionality to another function without changing its original code
def decorator(func):

    def wrapper():
        print("Starting function...")
        func()
        print("Function finished!")

    return wrapper


@decorator
def greet():
    print("Hello!")


greet()
```

```python
#yeild
def demo():
    print("Starting")
    yield 1
    print("Resumed after first yield")
    yield 2
    print("Resumed after second yield")
    yield 3
    print("Generator is done")

gen = demo()   # NOTE: nothing prints yet! Function body hasn't run at all

print("About to call next() #1")
print(next(gen))

print("About to call next() #2")
print(next(gen))

print("About to call next() #3")
print(next(gen))
```

```python
# *arags and **kwargs with sequence
def full_example(a, b, *args, c=10, **kwargs):
    print("a:", a)
    print("b:", b)
    print("args:", args)
    print("c:", c)
    print("kwargs:", kwargs)

full_example(1, 2, 3, 4, 5, c=99, x=100, y=200)
```

```python
#unpacking list tuple using *
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add(*nums))   # same as add(1, 2, 3)
```

```python
#unpacking  dict with **
def greet(name, role):
    print(f"{name} works as a {role}")

info = {"name": "Shraddha", "role": "Senior Executive"}
greet(**info)   # same as greet(name="Shraddha", role="Senior Executive")
```

```python
#lambda
square = lambda x: x * x
add=lambda a,b:a+b
print(add(1,2))
print(square(5))
```

```python
#map
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x * x, numbers)

print(result)
print(list(result))
```

```python
#filter
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

```python
#functools.reduce
from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda a, b: a + b, numbers)
print(result)  # 15
```

```python
#functools.partial
from functools import partial

def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)

print(square(5))  # 25  (base=5, exponent=2 pre-filled)
print(cube(5))     # 125 (base=5, exponent=3 pre-filled)
```

```python
#functools.wrap
from functools import wraps

def my_decorator(func):
    @wraps(func)  
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
```

```python
#itertools.groupby

from itertools import groupby
data = ["apple", "ant", "banana", "berry", "cherry", "avocado"]
data_sorted = sorted(data, key=lambda word: word[0])
print(data_sorted)
# ['apple', 'ant', 'avocado', 'banana', 'berry', 'cherry']

grouped = groupby(data_sorted, key=lambda word: word[0])
for key, group in grouped:
    print(key, list(group))
```

```python
#itertools.chain

from itertools import chain

scores1 = [85, 90]
scores2 = [70, 75]
scores3 = [95]

combined = chain(scores1, scores2, scores3)
combined_list = list(combined)  

average = sum(combined_list) / len(combined_list)
print(combined_list)  # [85, 90, 70, 75, 95]
print(average)         # 83.0
```
