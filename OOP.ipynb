```python
#magic methods1
#class with __str__()
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)    #op:Tobias (36)

#without --str--()
class animal:
    def __init__(self,name):
      self.name=name
p2=animal("cat")
print(p2) #<__main__.animal object at 0x306bcfb60>

print(p1==(p2))
```

```python
#magic method2: __eq__()
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __eq__(self, other):
    return self.name == other.name and self.age == other.age

p1 = Person("Linus", 30)
p2 = Person("Linus", 30)

print(p1 == p2)
```

```python
#ClassMethods
class Dog:
    count = 0

    def __init__(self, name):
        self.name = name
        Dog.count += 1

    @classmethod
    def how_many(cls):   #class method
        return f"There are {cls.count} dogs."

Dog("Buddy")
Dog("Luna")
print(Dog.how_many())   # There are 2 dogs.
```

```python
#statismethod
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @staticmethod
    def is_valid_age(age):
        return 0 <= age <= 30

print(Dog.is_valid_age(5))    # True
print(Dog.is_valid_age(50))   # False
```

```python
#TypesOfMethods&Attributes
class Temperature:
    unit = "Celsius"          # class attribute

    def __init__(self, degrees):
        self.degrees = degrees     # instance attribute

    # Instance method: works with ONE object's data
    def describe(self):
        return f"{self.degrees}° {Temperature.unit}"

    # Class method: works with the CLASS, often makes new objects
    @classmethod
    def from_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)

    # Static method: related utility, needs no object or class data
    @staticmethod
    def is_freezing(c):
        return c <= 0

t1 = Temperature(25)
t2 = Temperature.from_fahrenheit(212)

print(t1.describe())                  # 25° Celsius
print(t2.describe())                  # 100.0° Celsius
print(Temperature.is_freezing(-3))    # True
```

```python
#inheritance
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name}, age {self.age}.")


class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def introduce(self):
        super().introduce()                      # do the normal intro...
        print(f"I teach {self.subject}.")        # ...then add more

class Student(Person):
    def __init__(self,name,age,std):
        super().__init__(name,age)
        self.std=std

    def intro(self):
        super().introduce()  
        print(f"I study in {self.std} standard")
         
t = Teacher("Mrs Khan", 42, "Maths")
t.introduce()

s=Student("shraddha Tole",25,12)
s.intro()

print(isinstance(Student,Teacher))
print(issubclass(Teacher,Person))
print(issubclass(Student,Person))
```

```python
#polymorphism
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def draw(self):
        print(f"Drawing a circle of radius {self.radius}")


class Square:
    def __init__(self, side):
        self.side = side

    def draw(self):
        print(f"Drawing a square of side {self.side}")


class Triangle:
    def __init__(self, base):
        self.base = base

    def draw(self):
        print(f"Drawing a triangle of base {self.base}")


def draw_all(shapes):
    for shape in shapes:
        shape.draw()        # each shape knows how to draw itself


draw_all([Circle(5), Square(3), Triangle(4)])
```

```python
#abstraction
from abc import ABC, abstractmethod

class Report(ABC):
    @abstractmethod
    def generate(self):
        # shared starting point
        return "=== REPORT ===\n"


class SalesReport(Report):
    def generate(self):
        header = super().generate()
        return header + "Sales are up 10%."


print(SalesReport().generate())
# === REPORT ===
# Sales are up 10%.
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @property
    @abstractmethod
    def wheels(self):
        ...

    def describe(self):
        return f"{type(self).__name__} with {self.wheels} wheels"


class Car(Vehicle):
    @property
    def wheels(self):
        return 4


class Twowheeler(Vehicle):
    @property
    def wheels(self):
        return 2


print(Car().describe())    # Car with 4 wheels
print(Twowheeler().describe())   # Bike with 2 wheels
```
