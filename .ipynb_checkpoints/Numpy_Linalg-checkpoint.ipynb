```python
import numpy as np
np.set_printoptions(suppress=True, precision=4)
A = np.array([[4., 1.], [2., 3.]])
b = np.array([9., 13.])
v = np.array([3., -4.])
```

```python
# @ vs * 
print(A @ A)    # [[18, 7], [14, 11]]
print(A * A)    # [[16, 1], [ 4,  9]]
```

```python
#inverse
print(np.linalg.inv(A))

#Determinent
print(np.linalg.det(A))
```

```python
#Ax=b solve func

x=np.linalg.solve(A,b)
print(x,A @ x)
```

```python
#cond:how nasty is a matrix 
#Condition number = (largest singular value)/(smallest singular value). It measures how much errors get amplified.
print(np.linalg.cond(A))     # 2.618  → healthy
```

```python
np.linalg.inv(np.array([[1., 2.], [2., 4.]]))   # LinAlgError: Singular matrix
```

```python
#norms
print(np.linalg.norm(A))
print(np.linalg.norm(A,ord=1))
print(np.linalg.norm(v,ord=1))
print(np.linalg.norm(v, np.inf))
```

```python
#matrix_rank: Rank = number of truly independent columns.
print(np.linalg.matrix_rank(A) )                             # 2  (full rank)
print(np.linalg.matrix_rank(np.array([[1.,2.],[2.,4.]])) )   # 1  (column 2 = 2 × column 1)
```

```python
#eigenval & eigenvector
vals, vecs = np.linalg.eig(A)
print(vals)      # [5.+0.j  2.+0.j]
print(vecs)      # columns: [0.7071, 0.7071] and [-0.4472, 0.8944]
print(A @ vecs[:, 0], vals[0] * vecs[:, 0])      # both [3.5355, 3.5355] ✅The eigenvectors are the COLUMNS (vecs[:, 0]), not the rows. This is a very common mistake.
```

```python
S = np.array([[2., 1.], [1., 2.]])
vals, vecs = np.linalg.eigh(S) # vals = [1., 3.]  (ascending!)
print(vals)
print(vecs)
```

```python
#SVD : singlular value decomposition

U, s, Vt = np.linalg.svd(A)
print(s)                        # [5.1167 1.9544]   singular values, largest first
print(U @ np.diag(s) @ Vt)      # [[4. 1.] [2. 3.]]  rebuilds A ✅
```
